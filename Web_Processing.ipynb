```python
# STEP 1: hapus duplikasi paragraf
import requests
import textwrap
from bs4 import BeautifulSoup

def scrape_detik_article(base_url, max_pages=5):
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    all_text = []

    for page in range(1, max_pages + 1):
        url = base_url if page == 1 else f"{base_url}?page={page}"
        response = requests.get(url, headers=headers)
        if response.status_code != 200:
            break

        soup = BeautifulSoup(response.text, 'html.parser')
        body = soup.find('div', class_='detail__body-text itp_bodycontent')
        if not body:
            break

        paragraphs = body.find_all('p')
        page_text = ' '.join(
            p.get_text().strip() for p in paragraphs
            if 'para_caption' not in p.get('class', [])
        )
        all_text.append(page_text)

        # cek apakah masih ada halaman berikutnya
        pagination_text = soup.get_text()
        if f"Halaman {page} dari {page}" in pagination_text:
            break

    return ' '.join(all_text)

url = "https://news.detik.com/berita/d-8675026/misi-paket-vape-narkoba-berujung-jule-direhabilitasi"
teks_berita = scrape_detik_article(url)
print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
import re

# STEP 2: hapus duplikasi paragraf
def remove_duplicate_text(text):
    n = len(text)
    half = n // 2
    if text[:half].strip() == text[half:].strip():
        return text[:half].strip()
    return text

teks_berita = remove_duplicate_text(teks_berita)

print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
# STEP 3: hapus placeholder video/embed
def remove_video_placeholder(text):
    return re.sub(r'\[.*?\]', '', text)

teks_berita = remove_video_placeholder(teks_berita)

print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
# STEP 4: lower casing
teks_berita = teks_berita.lower()

print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
# STEP 5: remove punctuation
import string
punc = string.punctuation
teks_berita = teks_berita.translate(str.maketrans('', '', punc))

print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
# STEP 6: rapikan spasi berlebih
teks_berita = re.sub(r'\s+', ' ', teks_berita).strip()

print(textwrap.fill(teks_berita, width=80))
print(f"\nJumlah karakter: {len(teks_berita)}")
```

```python
# STEP 7: stemming
from mpstemmer import MPStemmer
stemmer = MPStemmer()
teks_stem = stemmer.stem_kalimat(teks_berita)

print(textwrap.fill(teks_stem, width=80))
print(f"\nJumlah karakter: {len(teks_stem)}")
```

```python
# STEP 8: remove stopwords
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
sw = set(stopwords.words('indonesian'))
teks_final = " ".join([w for w in teks_stem.split() if w not in sw])

print(textwrap.fill(teks_final, width=80))
print(f"\nJumlah karakter: {len(teks_final)}")
```

```python
#STEP 9: Visualisasi data teks
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(width=1000, height=500, background_color='white', max_words=500).generate(teks_final)

plt.figure(figsize=(7, 4))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
